```python
import pandas as pd
import numpy as np
import random
import string

df_ventas = pd.read_csv('sales.csv')
df_ventas.head(3)
```

```python
def generar_codigos(cantidad):
  codigos = set()
  while len(codigos) < cantidad:
    letras = ''.join(random.choices(string.ascii_uppercase, k=4))
    numeros = ''.join(random.choices(string.digits, k=4))
    codigos.add(letras + '-' + numeros)
  return list(codigos)
```

```python
# Asignación de Códigos únicos
codigos_ventas = generar_codigos(len(df_ventas))
df_ventas['Unnamed: 0'] = codigos_ventas
df_ventas.head(3)
```

```python
# Cambio de nombre
df_ventas.rename(columns={'Unnamed: 0': 'code_sales'}, inplace=True)
df_ventas.head(3)
```

```python
df_ventas_importantes = df_ventas[
    (df_ventas['sum_total'] > 500000.00) &
     (df_ventas['store_id'] == 4)
]
df_ventas_importantes.head(3)
```

## Ordenar y Agrupar Dataframes

```python
df_ventas.head(5)
```

```python
# Ordenar dataframes
df_ventas_ordenado = df_ventas.sort_values(
    by=['date', 'quantity'],
    ascending=False
)
df_ventas_ordenado.head(5)
```

```python
# Agrupar por columnas
df_ventas_agrupado = df_ventas.groupby(
    by=['date', 'store_id']
)[
    ['sum_total', 'quantity']
].sum()
df_ventas_agrupado
```

## Fusionar Dataframes

```python
df_productos = pd.read_csv('catalog.csv')

# Tabla de productos
df_productos.drop(columns=['Unnamed: 0'], inplace=True)
df_productos.head(5)
```

```python
# Tabla de Ventas
df_ventas.head(5)
```

```python
# Reconocer llaves foráneas
df_productos['item_id'].duplicated().sum(), df_ventas['item_id'].duplicated().sum()
```

```python
# Tablas unidas
df_ventas_descriptiva = df_ventas.merge(
    df_productos,
    on='item_id',
    how='left'
)
df_ventas_descriptiva.head(5)
```

```python
# Tabla unida y agrupada
df_ventas_agrupado = df_ventas_descriptiva.groupby(
    by=['item_type', 'store_id']
)[[
    'sum_total', 'quantity'
]].sum()

df_ventas_agrupado
```

## Combinar Dataframes

```python
# Combinación basado en índices, con prefijos en columnas de nombre repetido
df_ventas_combinado = df_ventas.join(df_productos, lsuffix='_left', rsuffix='_right')
df_ventas_combinado.head(5)
```

## Concatenar Dataframes

```python
df_ventas_online = pd.read_csv('online.csv')

df_ventas_online.drop(columns=['Unnamed: 0'], inplace=True)
df_ventas_online.head(5)
```

```python
df_ventas_offline = pd.read_csv('sales.csv')

df_ventas_offline.drop(columns=['Unnamed: 0'], inplace=True)
df_ventas_offline.head(5)
```

```python
# Consolidar ventas
df_ventas_consolidado = pd.concat(
    [df_ventas_offline, df_ventas_online],
    keys=['offline', 'online']
)
df_ventas_consolidado.head(5)
```

```python
df_ventas_offline.shape, df_ventas_online.shape, df_ventas_consolidado.shape
```

```python
# Consolidad opuesto (para este caso, no sirve)
df_ventas_consolidado_opuesto = pd.concat([df_ventas_offline, df_ventas_online], axis=1)
df_ventas_consolidado_opuesto.head(5)
```

## Datos de Tiempo

```python
# Generación de periodos de tiempo
df_ventas_consolidado['consolidation_time'] = pd.date_range(
    start='2026-01-01 00:00:00',
    periods=len(df_ventas_consolidado),
    freq='7s',
)
df_ventas_consolidado
```

```python
# Cambio de tipo de dato
df_ventas_consolidado['date'] = pd.to_datetime(
    df_ventas_consolidado['date'],
    format='%Y-%m-%d'
)
df_ventas_consolidado.info()
```

```python
quantity_enteros = (df_ventas_consolidado['quantity'] % 1 == 0).sum()
quantity_flotantes = (df_ventas_consolidado['quantity'] % 1 != 0).sum()

quantity_enteros, quantity_flotantes
```

```python
# Cálculos con series de tiempo
df_ventas_consolidado['consolidation_time'] = df_ventas_consolidado[
    'consolidation_time'
] + pd.Timedelta(seconds=1)
df_ventas_consolidado.head(5)
```

```python
# Exportación de tablas consolidadas (Retail)
#   Por su tamaño al exportar, hago las restricciones: cantidades solo enteras (1M de registros menos)
#   y otra: solo ventas totales mayores a 1000 dólares (+6M de registros menos)

df_ventas_consolidado = df_ventas_consolidado[df_ventas_consolidado['quantity'] % 1 == 0]
df_ventas_consolidado['quantity'] = df_ventas_consolidado['quantity'].astype(int)

df_ventas_consolidado = df_ventas_consolidado[df_ventas_consolidado['sum_total'] >= 1000]
df_ventas_consolidado
```

```python
df_ventas_consolidado_descriptiva = df_ventas_consolidado_descriptiva[
    df_ventas_consolidado_descriptiva['quantity'] % 1 == 0
]
df_ventas_consolidado_descriptiva['quantity'] = df_ventas_consolidado_descriptiva['quantity'].astype(int)

df_ventas_consolidado_descriptiva = df_ventas_consolidado_descriptiva[
    df_ventas_consolidado_descriptiva['sum_total'] >= 1000
]
df_ventas_consolidado_descriptiva
```

```python
df_ventas_consolidado.to_excel(
    'retail_sales_consolidate.xlsx',
    sheet_name='consolidate',
    index=False
)
df_ventas_consolidado_descriptiva.to_excel(
    'retail_sales_consolidate.xlsx',
    sheet_name='consolidate_descriptive',
    index=False
)
```

```python
# Exportación de tablas complementarias (Retail)
# De igual forma se hacen las restricciones, excepto a la tabla agrupada

def filtrado_ventas(df):
  df = df[df['quantity'] % 1 == 0]
  df['quantity'] = df['quantity'].astype(int)
  df = df[df['sum_total'] >= 1000]
  return df

df_ventas_offline = filtrado_ventas(df_ventas_offline)
df_ventas_online = filtrado_ventas(df_ventas_online)
df_ventas_ordenado = filtrado_ventas(df_ventas_ordenado)
df_ventas_descriptiva = filtrado_ventas(df_ventas_descriptiva)

[
    df_ventas_offline.shape,
    df_ventas_online.shape,
    df_ventas_agrupado.shape,
    df_ventas_ordenado.shape,
    df_ventas_descriptiva.shape
]
```

```python
with pd.ExcelWriter('retail_sales.xlsx') as writer:
  df_ventas_offline.to_excel(writer, sheet_name='sales_off', index=False)
  df_ventas_online.to_excel(writer, sheet_name='sales_on', index=False)
  df_ventas_agrupado.to_excel(writer, sheet_name='sales_off_resume', index=False)
  df_ventas_ordenado.to_excel(writer, sheet_name='sales_off_ordered', index=False)
  df_ventas_descriptiva.to_excel(writer, sheet_name='sales_off_info', index=False)
```
