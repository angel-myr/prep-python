## Datos faltantes

```python
bancarios = np.array([1.85, 2.13, np.nan, 4.25, 5.08])
bancarios
```

```python
np.isnan(bancarios)
```

```python
np.mean(bancarios)
```

```python
# Funciones para arreglos con vacíos
np.nanmean(bancarios)
```

```python
# Sustituir por valores válidos
bancarios_validos = np.where(
    np.isnan(bancarios),
    np.nanmean(bancarios),
    bancarios
)

bancarios_validos
```

```python
# Eliminar elementos no válidos
bancarios_validos = bancarios[~np.isnan(bancarios)]
bancarios_validos
```

## Exportación e Importación de Datos

```python
pagos = np.genfromtxt('Pagos.csv', delimiter=',')
pagos
```

```python
pagos = np.genfromtxt(
    'Pagos.csv',
    delimiter=',',
    filling_values=0,
    skip_header=1,
    dtype = int
)
pagos
```

```python
# Eliminar columnas con puros ceros
pagos_numericos = pagos[:, ~np.all(pagos == 0, axis=0)]
pagos_numericos
```

```python
# Exportar
np.savetxt(
    'pagos_numericos.csv',
    pagos_numericos,
    delimiter=';',
    fmt='%d'
)
```
