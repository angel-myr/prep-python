## Limpieza de Datos

```python
df_ventas.info()
```

```python
# Cantidad nulos
df_ventas.isnull().sum()
```

```python
# Eliminar nulos
df_ventas_filtro = df_ventas.dropna()
df_ventas_filtro.info()
```

```python
# Nulos rellenados
valores_relleno = {"Time": '00:00:00', "Quantity": df_ventas['Quantity'].mean()}
df_ventas_rellenados = df_ventas.fillna(valores_relleno)
df_ventas_rellenados.info()
```

```python
# Cambio de tipo
df_ventas_rellenados["Quantity"] = df_ventas_rellenados['Quantity'].astype(int)
df_ventas_rellenados.info()
```

```python
# Eliminar duplicados
df_ventas_unicos = df_ventas.drop_duplicates()
df_ventas_unicos.info()
```

```python
# Eliminar duplicados parciales
df_ventas_unicos = df_ventas.drop_duplicates(subset='Invoice ID')
df_ventas_unicos.head(3)
```

## Filtrado de Series

```python
serie_totales = pd.Series(df_ventas['Total'])
serie_totales.head(5)
```

```python
# Filtrado de Series Numeral
filtro = serie_totales > serie_totales.mean()
serie_filtrada = serie_totales[filtro]
serie_filtrada, serie_filtrada.count(), serie_totales.count(), serie_totales.mean()
```

```python
# Filtrado de Series Cadenas
serie_ciudad = pd.Series(df_ventas['City'])
filtro = serie_ciudad.str.contains('M|m')
serie_filtrada = serie_ciudad[filtro]
serie_filtrada, serie_filtrada.count(), serie_ciudad.count()
```

## Agregaciones de Series

```python
promedio_ventas = serie_totales.mean()
suma_ventas = serie_totales.sum()
minimo_ventas = serie_totales.min()
maximo_ventas = serie_totales.max()
varianza_ventas = serie_totales.std()

promedio_ventas, suma_ventas, minimo_ventas, maximo_ventas, varianza_ventas
```
